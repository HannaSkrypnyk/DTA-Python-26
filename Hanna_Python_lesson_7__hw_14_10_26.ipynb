1. Створи змінну `aqi = 85` — індекс якості повітря. Виведи оцінку якості повітря:
- від 0 до 50 — `"Good"`;
- від 51 до 100 — `"Moderate"`;
- від 101 до 150 — `"Unhealthy for sensitive groups"`;
- від 151 до 200 — `"Unhealthy"`;
- понад 200 — `"Hazardous"`.

Перевір програму для значень 30, 85, 150 і 240

```python
aqi = 85
if aqi <= 50:
  print("Good")
elif aqi <= 100:
  print("Moderate")
elif aqi <= 150:
  print("Unhealthy for sensitive groups")
elif aqi <= 200:
  print("Unhealthy")
else:
  print("Hazardous")
```

4. Створи змінну `kwh = 320` — спожита електроенергія за місяць у кВт·год. Оплата розраховується ступенево:
- перші 100 кВт·год — по 2.64 грн;
- наступні 150 кВт·год (від 101 до 250) — по 3.20 грн;
- усе понад 250 кВт·год — по 4.32 грн.

Кожна частина споживання оплачується за своїм тарифом. Виведи суму до сплати, округлену до 2 знаків.  
Перевір програму для значень 80, 180 і 320

```python
kwh = 320
if kwh <= 100:
  cost = kwh * 2.64
elif kwh <= 250:
  cost = 100 * 2.64 + (kwh - 100) * 3.20
else:
  cost = 100 * 2.64 + 150 * 3.20 + (kwh - 250) * 4.32
print(round(cost, 2), "грн.")
```

5. Створи змінні `a = 3`, `b = 4`, `c = 5` — довжини сторін трикутника. Виведи тип трикутника:
- якщо сума будь-яких двох сторін не більша за третю, такого трикутника не існує — `"Not a triangle"`;
- усі сторони рівні — `"Equilateral"`;
- дві сторони рівні — `"Isosceles"`;
- трикутник прямокутний (квадрат однієї сторони дорівнює сумі квадратів двох інших) — `"Right"`;
- в інших випадках — `"Scalene"`.

Перевір програму для трійок (3, 4, 5), (5, 3, 4), (2, 2, 2), (5, 5, 8), (1, 2, 3) і (4, 6, 7)

```python
a = 1
b = 2
c = 3
if a + b <= c or a + c <= b or b + c <= a:
  print("Not a triangle")
elif a == b == c:
  print("Equilateral")
elif a == b or b == c or a == c:
  print("Isosceles")
elif a**2 == b**2 + c**2 or b**2 == a**2 + c**2 or c**2 == a**2 + b**2:
  print("Right")
else:
  print("Scalene")
```

7. Створи змінну `number = 90517`. Порахуй суму цифр числа та кількість цифр у ньому.  
Не перетворюй число на рядок. Перевір програму також для числа 7

```python
number = 90517

n = number          # рабочая копия, чтобы не потерять оригинал
total = 0
count = 0

while True:
    digit = n % 10
    total += digit
    count += 1
    n = n // 10
    if n == 0:
      break

print("Число:", number)
print("Сума цифр:", total)
print("Кількість цифр:", count)
```

```python
number = 90517
n = number
total = 0
count = 0

while n > 0:
    n, digit = divmod(n, 10)
    total += digit
    count += 1

print("Число:", number)
print("Сума цифр:", total)
print("Кількість цифр:", count)
```
