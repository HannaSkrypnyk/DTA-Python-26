## Завдання 1. Перевірка та очищення даних

Переглянь усі поїздки та розподіли їх:
- скасовані поїздки;
- **некоректні** завершені поїздки: відстань дорівнює 0 або менше — причина `"invalid distance"`; оцінка не є цілим числом від 1 до 5 — причина `"invalid rating"`. Якщо в поїздці кілька проблем, вказуй лише першу з них у цьому порядку;
- **коректні** завершені поїздки — збережи їх в окремий список `valid_rides`, з ним ти працюватимеш далі.

Виведи загальну кількість поїздок, кількість скасованих, кількість коректних, а також номер і причину для кожної некоректної поїздки

```python
valid_rides = []
cancelled_count = 0
invalid = []

for ride in rides:
  if ride['status'] == 'cancelled':
    cancelled_count += 1
    continue

  distance = ride["distance"]
  rating = ride["rating"]

  if distance <= 0:
    invalid.append((ride["id"], "invalid distance"))
  elif not isinstance(rating, int) or rating < 1 or rating > 5:
    invalid.append((ride["id"], "invalid rating"))
  else:
    valid_rides.append(ride)

print("Усього поїздок:", len(rides))
print("Скасованих:", cancelled_count)
print("Коректних:", len(valid_rides))
print()
for ride_id, reason in invalid:
    print(ride_id, "—", reason)

```

## Завдання 2. Розрахунок вартості поїздок

Для кожної поїздки з `valid_rides` розрахуй вартість за тарифом:
- посадка — 50 грн плюс 12 грн за кожен кілометр;
- якщо поїздка тривала понад 30 хвилин — додатково 3 грн за кожну хвилину понад 30;
- нічний тариф (поїздка почалася з 22:00 до 05:59 включно) — уся вартість збільшується на 25%;
- мінімальна вартість поїздки — 80 грн: якщо розрахована сума менша, пасажир платить 80 грн.

Округли вартість до 2 знаків і додай її до словника кожної поїздки з ключем `"price"`. Виведи для кожної поїздки номер, водія, відстань і вартість

```python
for ride in valid_rides:
  price = 50 + 12 * ride["distance"]

  if ride["duration"] > 30:
    price += 3 * (ride["duration"] - 30)
  if ride["hour"] >= 22 or ride["hour"] <= 5:
    price *= 1.25
  if price < 80:
    price = 80

  ride["price"] = round(price, 2)
  print(ride["id"], ride["driver"], ride["distance"], ride["price"])
print()
print(rides)
```

## Завдання 3. Статистика по водіях

На основі `valid_rides` створи словник `driver_stats`, у якому для кожного водія зберігаються:
- кількість поїздок;
- виручка (сума вартостей його поїздок, округлена до 2 знаків);
- середній рейтинг, округлений до 1 знака.

Виведи статистику кожного водія в одному рядку

```python
driver_stats = {}

for ride in valid_rides:
  name = ride["driver"]
  if name not in driver_stats:
    driver_stats[name] = {"count": 0, "revenue": 0, "ratings": []}

    driver_stats[name]["count"] += 1
    driver_stats[name]["revenue"] += ride["price"]
    driver_stats[name]["ratings"].append(ride["rating"])

for name, stats in driver_stats.items():
  stats["revenue"] = round(stats["revenue"], 2)
  stats["avg_rating"] = round(sum(stats["ratings"]) / len(stats["ratings"]), 1)
  print(name, stats["count"], stats["revenue"], stats["avg_rating"])

print()
print(driver_stats)
```

## Завдання 4. Статуси та бонуси водіїв

На основі `driver_stats` визнач для кожного водія статус і бонус (середній рейтинг береться округленим до 1 знака):
- рейтинг від 4.8 **і** щонайменше 3 поїздки — статус `"Gold"`, бонус 10% від виручки;
- рейтинг від 4.5 — статус `"Silver"`, бонус 5% від виручки;
- рейтинг нижче 4.0 — статус `"Warning"`, бонусу немає;
- в інших випадках — статус `"Standard"`, бонусу немає.

Округли бонус до 2 знаків і додай статус та бонус до статистики водія. Виведи імʼя, статус і бонус кожного водія, а наприкінці — загальну суму бонусів

```python
total_bonus = 0

for name, stats in driver_stats.items():
  rating = stats["avg_rating"]

  if rating >= 4.8 and stats["count"] >= 3:
    status = "Gold"
    bonus = round(stats["revenue"] * 0.10, 2)
  elif rating >= 4.5:
    status = "Silver"
    bonus = round(stats["revenue"] * 0.05, 2)
  elif rating < 4.0:
    status = "Warning"
    bonus = 0
  else:
    status = "Standard"
    bonus = 0

  stats["status"] = status
  stats["bonus"] = bonus
  total_bonus += bonus
  print(name, status, bonus)

print("Загальна сума бонусів:", round(total_bonus, 2))
print()
print(driver_stats)
```

## Завдання 5. Підсумковий звіт за день

Виведи звіт, у якому кожен показник займає окремий рядок:
- загальна виручка за день;
- кількість коректних поїздок і середня вартість поїздки (округли до 2 знаків);
- частка скасованих поїздок від усіх поїздок у журналі, % (округли до 1 знака);
- частка поїздок, оплачених карткою, серед коректних поїздок, % (округли до 1 знака);
- найзавантаженіша година — година, на яку припало найбільше коректних поїздок, і кількість поїздок у цю годину;
- найдовша поїздка: номер, водій і відстань;
- водій з найбільшою виручкою;
- загальна сума бонусів водіям.

Не використовуй функції `max()` і `min()`

```python
# 1. Загальна виручка
total_revenue = 0
for ride in valid_rides:
  total_revenue += ride["price"]

# 2. Середня вартість
avg_price = total_revenue / len(valid_rides)

# 3. Частка скасованих (від усіх поїздок у журналі)
cancelled_share = cancelled_count / len(rides) * 100

# 4. Частка оплат карткою серед коректних
card_count = 0
for ride in valid_rides:
  if ride["payment"] == "card":
    card_count += 1
card_share = card_count / len(valid_rides) * 100

# 5. Найзавантаженіша година
hours = {}
for ride in valid_rides:
  hours[ride["hour"]] = hours.get(ride["hour"], 0) + 1

busy_hour = None
busy_count = 0
for hour, count in hours.items():
  if busy_hour is None or count > busy_count:
    busy_hour = hour
    busy_count = count

# 6. Найдовша поїздка
longest = None
for ride in valid_rides:
  if longest is None or ride["distance"] > longest["distance"]:
    longest = ride

# 7. Водій з найбільшою виручкою
top_driver = None
top_revenue = 0
for name, stats in driver_stats.items():
  if top_driver is None or stats["revenue"] > top_revenue:
    top_driver = name
    top_revenue = stats["revenue"]

print("Загальна виручка:", round(total_revenue, 2))
print("Коректних поїздок:", len(valid_rides), "| середня вартість:", round(avg_price, 2))
print("Скасовано:", round(cancelled_share, 1), "%")
print("Оплата карткою:", round(card_share, 1), "%")
print("Найзавантаженіша година:", busy_hour, "—", busy_count, "поїздок")
print("Найдовша поїздка:", longest["id"], longest["driver"], longest["distance"], "км")
print("Найбільша виручка:", top_driver, round(top_revenue, 2))
print("Сума бонусів:", round(total_bonus, 2))
```
