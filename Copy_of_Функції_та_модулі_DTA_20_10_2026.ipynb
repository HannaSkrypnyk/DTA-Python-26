# Тема 1. Оголошення функцій
### Funktionen definieren

- будувати розрахунок з кількох маленьких функцій.

## 1.1 Що таке функція

Ви вже користувалися функціями — вбудованими у Python:

````python
sales = [1240, 980, 1530]

print(len(sales))   # len — функція: рахує кількість елементів
print(sum(sales))   # sum — функція: рахує суму
print(max(sales))   # max — функція: знаходить найбільше значення
````

**Функція** — це іменований «рецепт»: ми один раз описуємо, що робити, а потім просто називаємо рецепт і отримуємо результат.

💡 **Аналогія для аналітика:** функція схожа на формулу в Excel. `=AVG(A1:A10)` отримує дані на вході й повертає результат у клітинку. Але в Python Ви можете створювати **власні** «формули» для будь-яких розрахунків.

**Навіщо створювати власні функції?**
- **Не повторювати код.** Розрахунок пишемо один раз, а використовуємо скільки завгодно.
- **Менше помилок.** Якщо формула зміниться, виправляти потрібно лише в одному місці.
- **Код легше читати.** `conversion_rate(orders, visitors)` зрозуміліше, ніж набір чисел і знаків.

## 1.2 Перша функція: крок за кроком

````python
def say_hello():
    print("Вітаємо в аналітиці!")
    print("Сьогодні вивчаємо функції.")
````

Запустіть клітинку вище. **Нічого не з'явилося?** Так і має бути! Ми лише **описали** функцію (*оголосили*), але ще не **викликали** її.

Розберемо запис:

| Частина | Що означає |
|---|---|
| `def` | від англ. *define* — «визначити»: тут починається функція |
| `say_hello` | назва функції (правила ті самі, що для змінних) |
| `()` | дужки: тут будуть вхідні дані (поки порожньо) |
| `:` | двокрапка: «далі — тіло функції» |
| відступ 4 пробіли | усі рядки з відступом належать функції |

Тепер **викличемо** функцію — напишемо її назву з дужками:

````python
say_hello()
say_hello()   # функцію можна викликати скільки завгодно разів
````

## 1.3 Параметри: передаємо дані у функцію

Щоб функція працювала з різними даними, ми додаємо **параметри** — назви в дужках. Під час виклику на їхнє місце підставляються конкретні значення — **аргументи**.

````python
def show_revenue(orders, avg_check):
    revenue = orders * avg_check
    print("Виручка:", revenue, "грн")


show_revenue(120, 350)   # orders = 120, avg_check = 350
show_revenue(95, 410)    # orders = 95,  avg_check = 410
````

**Що відбувається під час виклику `show_revenue(120, 350)`:**
1. Python знаходить функцію `show_revenue`.
2. Підставляє значення: `orders = 120`, `avg_check = 350` (за порядком).
3. Виконує тіло функції: `revenue = 120 * 350 = 42000`.
4. Друкує результат.

| Термін | Де | Приклад |
|---|---|---|
| **параметр** | в описі функції (`def`) | `orders`, `avg_check` |
| **аргумент** | під час виклику | `120`, `350` |

````python
def show_avg_score(score, count=10):
    # avg_score = sum(score) / count
    avg_score = score / count
    print("Avg score:", avg_score)

show_avg_score(120)
show_avg_score(120, 20)
show_avg_score(score=120, count=10)
show_avg_score(count=10, score=120)
````

````python
def show_avg_score(score: float|int, count: int=10) -> float:
    """Func show avg score"""

    # avg_score = sum(score) / count  # score: list
    avg_score = score / count
    print("Avg score:", avg_score)

show_avg_score(120)
show_avg_score(120, 20)
show_avg_score(score=120, count=10)
show_avg_score(count=10, score=120)
````

## 1.4 `return`: отримуємо результат

Функція `show_revenue` лише **друкує** виручку. Але аналітику зазвичай потрібно **використати** число далі: скласти, порівняти, записати у звіт. Для цього є `return` — він **повертає** результат туди, звідки функцію викликали.

````python
def revenue(orders, avg_check):
    return orders * avg_check


monday = revenue(120, 350)    # результат функції потрапляє в змінну
tuesday = revenue(95, 410)

print("Понеділок:", monday)
print("Вівторок:", tuesday)
print("Разом за два дні:", monday + tuesday)
````

**`print` чи `return`?**

| | `print(...)` | `return ...` |
|---|---|---|
| Що робить | показує значення на екрані | віддає значення назад у програму |
| Чи можна рахувати з результатом далі | ❌ ні | ✅ так |
| Аналогія | написати число на дошці | формула Excel поклала число в клітинку |

Якщо у функції немає `return`, вона повертає `None` — спеціальне значення «нічого»:

````python
def total_print(a, b):
    print(a + b)        # лише показує


def total_return(a, b):
    return a + b        # повертає


x = total_print(2, 3)   # на екрані з'явиться 5, але в x нічого не потрапить
y = total_return(2, 3)

print("x =", x)
print("y =", y)
````

**Важливо:** `return` одразу **завершує** функцію. Тому його зручно використовувати для перевірок на початку функції:

````python
def average_check(revenue, orders):
    if orders == 0:
        return 0                  # замовлень немає → одразу виходимо, ділити на нуль не будемо
    return revenue / orders       # сюди дійдемо, лише якщо orders не 0


print(average_check(42000, 120))
print(average_check(0, 0))
````

## 1.5 Docstring: опис функції

**Docstring** — це текст у потрійних лапках одразу під рядком з `def`. Він пояснює, що робить функція. Через кілька тижнів Ви (або Ваші колеги) швидко зрозумієте, як нею користуватися.

Прочитати опис можна за допомогою `help()`. А в Colab він з'являється й тоді, коли Ви наводите курсор на назву функції.

````python
def margin(revenue, cost):
    """Рахує маржу у відсотках.

    revenue — виручка, cost — собівартість.
    """
    return (revenue - cost) / revenue * 100


help(margin)
print(margin(1000, 650))
````

## 1.6 Де «живуть» змінні: область видимості

- Змінні, створені **всередині** функції, — **локальні**. Вони існують, лише поки функція працює. Це як чернетка: після розрахунку її викидають.
- Змінні, створені **поза** функціями, — **глобальні**. Функція може їх **читати**.

````python
tax_rate = 0.2                    # глобальна змінна


def price_with_tax(price):
    tax = price * tax_rate        # tax — локальна; tax_rate функція читає ззовні
    return price + tax


print(price_with_tax(100))
# print(tax)   # ❌ NameError: поза функцією змінної tax не існує. Приберіть # і перевірте самі.
````

````python
tax_rate = 0.2                    # глобальна змінна


def price_with_tax(price):
    global tax
    tax = price * tax_rate        # tax — локальна; tax_rate функція читає ззовні

    return price + tax


print(price_with_tax(100))
print(tax)   # ❌ NameError: поза функцією змінної tax не існує. Приберіть # і перевірте самі.
````

✅ **Головне правило:** дані потрапляють у функцію **через параметри**, а результат виходить **через `return`**. Тоді функція працює однаково, де б Ви її не викликали.

## 1.7 Функції працюють разом

Великий розрахунок зручно розбити на маленькі функції. Кожна робить **одну** річ, а потім одна функція викликає іншу.

````python
def subtotal(price, quantity):
    return price * quantity


def apply_discount(amount, percent):
    return amount * (1 - percent / 100)


def final_price(price, quantity, discount_percent):
    amount = subtotal(price, quantity)                 # крок 1: сума без знижки
    return apply_discount(amount, discount_percent)    # крок 2: застосовуємо знижку


print(final_price(250, 4, 10))
````

**Покроково для `final_price(250, 4, 10)`:**
1. `subtotal(250, 4)` → `1000`
2. `apply_discount(1000, 10)` → `1000 * 0.9` → `900.0`
3. `final_price` повертає `900.0`

## ⚠️ Типові помилки

| Що бачите | Причина | Як виправити |
|---|---|---|
| `SyntaxError: expected ':'` | забули двокрапку після `def ...()` | `def revenue(orders, avg_check):` |
| `IndentationError` | немає відступу в тілі функції | 4 пробіли (або Tab) перед кожним рядком тіла |
| `NameError: name 'revenue' is not defined` | функцію викликали до того, як запустили клітинку з `def` | спершу запустіть клітинку з описом функції |
| у результаті `None` | у функції немає `return` | замініть `print(...)` на `return ...` |
| `TypeError: ... missing 1 required positional argument` | передали менше аргументів, ніж потрібно | передайте всі аргументи в дужках |

### 🧭 Завдання 1.3 · Конверсія з перевіркою на нуль
*Розібране завдання: є підказка й пояснення.*

**Конверсія** — частка відвідувачів сайту, які зробили замовлення, у відсотках.

Напишіть функцію `conversion_rate(orders, visitors)`:
- якщо `visitors` дорівнює 0, функція повертає `0`;
- інакше повертає `orders / visitors * 100`, округлене до 2 знаків після коми.

<details>
<summary>💡 <b>Підказка</b></summary>

**Крок 1.** Спершу перевірка: `if visitors == 0:` і під нею з відступом `return 0`.

**Крок 2.** Далі, вже без `if`: `return round(orders / visitors * 100, 2)`.

</details>

````python
def conversion_rate(orders, visitors):
    # 🔧 крок 1: перевірка на нуль
    if visitors == 0:  # if not visitors:
        return 0
    # 🔧 крок 2: розрахунок конверсії
    return round(orders / visitors * 100, 2)


print(conversion_rate(25, 1000))
print(conversion_rate(1, 3))
print(conversion_rate(10, 0))
````

</details>

## 2.1 Позиційні аргументи: важливий порядок

Коли ми пишемо аргументи просто через кому, Python підставляє їх **за порядком**: перший аргумент — у перший параметр, другий — у другий і так далі.

````python
def product_card(name, price, stock):
    return f"{name}: {price} грн, на складі {stock} шт."


print(product_card("Кава зернова", 389, 12))   # правильний порядок
print(product_card("Кава зернова", 12, 389))   # переплутали порядок — і результат хибний!
````

Python не помітив помилки: обидва числа — допустимі значення. Тому з позиційними аргументами важливо **пам'ятати порядок**.

## 2.2 Іменовані аргументи: передаємо за назвою

Аргумент можна передати **за назвою параметра**: `назва=значення`. Тоді порядок не важливий, а код читається як інструкція.

````python
print(product_card(name="Кава зернова", price=389, stock=12))
print(product_card(stock=12, price=389, name="Кава зернова"))   # той самий результат

# Можна поєднувати: спочатку позиційні, потім іменовані
print(product_card("Кава зернова", stock=12, price=389))
````

⚠️ Після іменованого аргументу позиційний писати **не можна**: `product_card(name="Кава", 389, 12)` дасть `SyntaxError`.

## 2.3 Значення за замовчуванням

Деякі параметри майже завжди мають одне значення. Наприклад, ставка ПДВ в Україні — 20 %. Таке значення можна вказати **за замовчуванням**, і тоді аргумент можна не передавати.

````python
def price_with_vat(price, rate=0.2):
    return round(price * (1 + rate), 2)


print(price_with_vat(100))              # rate не передали → береться 0.2
print(price_with_vat(100, 0.07))        # передали іншу ставку
print(price_with_vat(100, rate=0.19))   # або за назвою — так зрозуміліше
````

**Правило:** параметри зі значенням за замовчуванням пишуть **після** обов'язкових.
- ✅ `def price_with_vat(price, rate=0.2):`
- ❌ `def price_with_vat(rate=0.2, price):` — `SyntaxError`

⚠️ **Зверніть увагу:** не використовуйте список чи словник як значення за замовчуванням (`def f(items=[])`). Такий список створюється **один раз** і «накопичує» дані між викликами. Пишіть `items=None`, а всередині функції: `if items is None: items = []`.

## 2.4 `*args`: будь-яка кількість аргументів

Іноді наперед невідомо, скільки значень передадуть у функцію. Згадайте Excel: у `=СУМ(A1; B5; C3)` можна передати скільки завгодно клітинок. У Python для цього є **зірочка** перед назвою параметра.

````python
def total_sales(*amounts):
    print("Усередині функції amounts =", amounts)
    return sum(amounts)


print(total_sales(120, 300, 75))
print(total_sales(500))
print(total_sales())
````

**Що відбувається:**
1. Зірочка `*` збирає **всі** передані значення в один кортеж.
2. Усередині функції `amounts` — це кортеж, наприклад `(120, 300, 75)`.
3. З ним працюємо як зі списком: `sum()`, `len()`, цикл `for`.

Назва `args` — лише традиція (від *arguments*); важлива саме зірочка.

## 2.5 `**kwargs`: будь-які іменовані аргументи

Дві зірочки збирають усі **іменовані** аргументи у **словник**. Це зручно, коли в кожного об'єкта може бути свій набір характеристик.

````python
def customer_profile(name, **details):
    print("Усередині функції details =", details)
    profile = {"name": name}
    profile.update(details)        # додаємо всі пари зі словника details
    return profile


print(customer_profile("Олена", city="Київ", orders=7))
print(customer_profile("Андрій"))
````

| Запис | Збирає | Тип усередині функції |
|---|---|---|
| `*args` | позиційні аргументи | кортеж `(...)` |
| `**kwargs` | іменовані аргументи | словник `{...}` |

## 2.6 Анотації типів: підписуємо, що очікуємо

**Анотації** (*type hints*) — це підказки, якого типу мають бути дані. Пишемо `параметр: тип`, а тип результату — після стрілки `->`.

````python
def average_order_value(revenue: float, orders: int) -> float:
    if orders == 0:
        return 0.0
    return revenue / orders


print(average_order_value(42000.0, 120))
help(average_order_value)    # анотації видно в описі функції
````

| Анотація | Значення |
|---|---|
| `int` | ціле число |
| `float` | дробове число |
| `str` | рядок |
| `bool` | `True` / `False` |
| `list[float]` | список дробових чисел |
| `dict[str, int]` | словник: ключі — рядки, значення — цілі числа |

Python **не перевіряє** анотації: це підказка для людей і для редактора коду. Але вони роблять функцію набагато зрозумілішою.

## 3.1 Функцію можна передати як значення

Назва функції **без дужок** — це сама функція. Назва **з дужками** — це її виклик, тобто результат.

````python
def add_fee(amount):
    return amount + 25


print(add_fee(100))   # з дужками: викликаємо → 125
print(add_fee)        # без дужок: сама функція

fee_function = add_fee            # кладемо функцію в змінну
print(fee_function(200))          # і викликаємо через нову назву → 225
````

Отже, функцію можна передати в **іншу** функцію — так само, як число чи список. Функцію, яка приймає іншу функцію, називають **функцією вищого порядку**.

````python
def apply_to_all(func, values):
    result = []
    for value in values:
        result.append(func(value))   # викликаємо передану функцію для кожного значення
    return result


print(apply_to_all(add_fee, [100, 200, 300]))
print(apply_to_all(len, ["кава", "чай"]))      # вбудовану функцію теж можна передати
````

## 3.2 Параметр `key`: сортуємо за своїм правилом

Ви вже знаєте `sorted()`. У нього є параметр `key`, у який передають **функцію**. Python викликає її для кожного елемента й сортує **за її результатами**.

````python
cities = ["Харків", "Київ", "Ужгород", "Львів"]

print(sorted(cities))            # за алфавітом
print(sorted(cities, key=len))   # за довжиною назви
````

**Як працює `sorted(cities, key=len)` — покроково:**

| Місто | `len(місто)` |
|---|---|
| Харків | 6 |
| Київ | 4 |
| Ужгород | 7 |
| Львів | 5 |

Python сортує за числами в правому стовпці: 4 → 5 → 6 → 7, тобто Київ, Львів, Харків, Ужгород.

## 3.3 `lambda`: функція в один рядок

Часто для `key` потрібна маленька функція, яку використовуємо лише один раз. Писати для неї повний `def` незручно. Для цього є **lambda**:

```
lambda аргумент: що повернути
```

````python
def double(x):
    return x * 2


double_lambda = lambda x: x * 2    # те саме, що функція double

print(double(5), double_lambda(5))
````

| `def` | `lambda` |
|---|---|
| `def double(x):`<br>`    return x * 2` | `lambda x: x * 2` |
| має назву | без назви («анонімна») |
| може мати багато рядків | лише **один** вираз |
| `return` пишемо | `return` **не** пишемо: результат виразу і є відповіддю |

💡 Lambda зазвичай не кладуть у змінну, а пишуть **одразу там, де вона потрібна**, наприклад у `key=`.

## 3.4 Сортування товарів: `sorted`, `max`, `min` з `key`

В аналітиці дані часто зберігаються як **список словників**: один словник — один товар. Щоб відсортувати товари за ціною, потрібно сказати Python: «для кожного товару бери значення `"price"`».

````python
products = [
    {"name": "Чайник", "price": 1450, "rating": 4.3},
    {"name": "Тостер", "price": 980, "rating": 4.7},
    {"name": "Блендер", "price": 2100, "rating": 4.5},
]

by_price = sorted(products, key=lambda p: p["price"])
for p in by_price:
    print(p["price"], p["name"])
````

**Як читати `key=lambda p: p["price"]`:** «для кожного товару `p` бери його ціну `p["price"]`».

**За спаданням** — додайте `reverse=True`:

````python
by_rating = sorted(products, key=lambda p: p["rating"], reverse=True)
for p in by_rating:
    print(p["rating"], p["name"])
````

`max()` і `min()` теж приймають `key`. Вони повертають **увесь** словник товару, а не лише ціну:

````python
cheapest = min(products, key=lambda p: p["price"])
best = max(products, key=lambda p: p["rating"])

print("Найдешевший:", cheapest)
print("Найкращий рейтинг:", best["name"])
````

## 3.5 `map`: застосувати функцію до кожного елемента

`map(функція, список)` викликає функцію для **кожного** елемента. Щоб побачити результат, обгортаємо його в `list()`.

````python
prices = [100, 250, 80]

with_vat = list(map(lambda p: p * 1.2, prices))
print(with_vat)

print(map(lambda p: p * 1.2, prices))   # без list() побачимо лише «об'єкт map», а не числа
````

## 3.6 `filter`: залишити лише потрібне

`filter(функція, список)` залишає лише ті елементи, для яких функція повернула `True`.

````python
prices = [1200, 450, 3100, 89, 760]

expensive = list(filter(lambda p: p > 1000, prices))
print(expensive)

top_products = list(filter(lambda p: p["rating"] >= 4.5, products))
for p in top_products:
    print(p["name"], p["rating"])
````

**Шпаргалка**

| Завдання | Код |
|---|---|
| відсортувати за полем | `sorted(items, key=lambda x: x["поле"])` |
| за спаданням | `sorted(items, key=..., reverse=True)` |
| знайти найбільший | `max(items, key=lambda x: x["поле"])` |
| змінити кожен елемент | `list(map(lambda x: ..., items))` |
| відібрати за умовою | `list(filter(lambda x: умова, items))` |

## ⚠️ Типові помилки

| Що бачите | Причина | Як виправити |
|---|---|---|
| `<map object at 0x...>` або `<filter object ...>` | забули `list()` | `list(map(...))` |
| `TypeError: 'int' object is not callable` | у `key` передали результат, а не функцію: `key=len()` | `key=len` (без дужок) |
| `SyntaxError` у lambda | написали `return` | `lambda x: x * 2`, без `return` |

### 🏁 Завдання 3.10 · Міні-проєкт: вітрина з фільтрами

Напишіть функцію для вітрини інтернет-магазину:

```
select_products(products, max_price, min_rating)
```

Вона повертає **список назв** товарів, які:
- коштують **не більше** `max_price`;
- мають рейтинг **не нижче** `min_rating`;
- відсортовані за ціною — від дешевих до дорогих.

**План — крок за кроком:**
1. `filter` — залиште товари з ціною `<= max_price`.
2. Ще один `filter` — залиште товари з рейтингом `>= min_rating`.
3. `sorted` — відсортуйте за ціною.
4. `map` — візьміть лише назви.
5. Поверніть результат як список.

Потім виконайте два «запити покупця»: `request_1` — до 2000 грн з рейтингом від 4.5, `request_2` — до 1000 грн з будь-яким рейтингом (від 0).

<details>
<summary>💡 <b>Підказка</b></summary>

Результат кожного кроку можна зберігати в окрему змінну:

```
step1 = filter(lambda p: p["price"] <= max_price, products)
step2 = filter(lambda p: p["rating"] >= min_rating, step1)
step3 = sorted(step2, key=lambda p: p["price"])
return list(map(lambda p: p["name"], step3))
```

</details>

````python
def select_products(products, max_price, min_rating):
    cheap_items = filter(lambda item: item['price'] < max_price, products)    # 🔧 кроки 1–5 з умови
    top_cheap_items = filter(lambda item: item['rating'] > min_rating, cheap_items)
    sorted_items = sorted(top_cheap_items, key=lambda item: item["price"])
    return list(map(lambda item: item["name"], sorted_items))



request_1 = select_products(PRODUCTS, 2000, 4.5)   # 🔧 до 2000 грн, рейтинг від 4.5
request_2 = select_products(PRODUCTS, 1000, 0)   # 🔧 до 1000 грн, будь-який рейтинг

print("Запит 1:", request_1)
print("Запит 2:", request_2)
````

| `%Y` | рік (4 цифри) | 2026 |
| `%m` | місяць (2 цифри) | 10 |
| `%d` | день (2 цифри) | 16 |

Формат має **точно** повторювати рядок, з усіма дефісами чи крапками: для `"2026-10-16"` — `"%Y-%m-%d"`, для `"16.10.2026"` — `"%d.%m.%Y"`.

````python
from datetime import datetime

order = datetime.strptime("2026-10-16", "%Y-%m-%d")
print(order)

order_2 = datetime.strptime("16.10.2026", "%d.%m.%Y")   # інший формат рядка → інший шаблон
print(order_2)
````

`.weekday()` повертає номер дня: `0` — понеділок, `1` — вівторок … `6` — неділя.

**Крок 4. Дата → гарний рядок**

`.strftime(формат)` робить навпаки: перетворює дату на текст у потрібному вигляді.

````python
print(order.strftime("%d.%m.%Y"))   # 16.10.2026
````

**Крок 5. Різниця дат і додавання днів**

Якщо відняти одну дату від іншої, отримаємо **проміжок часу** (`timedelta`). Кількість днів у ньому — `.days`. А щоб додати до дати кілька днів, використовуємо `timedelta(days=...)`.

````python
from datetime import datetime, timedelta

ordered = datetime.strptime("2026-10-05", "%Y-%m-%d")
delivered = datetime.strptime("2026-10-08", "%Y-%m-%d")

gap = delivered - ordered
print("Доставка тривала:", gap.days, "дн.")

expected = ordered + timedelta(days=5)
print("Очікувана дата:", expected.strftime("%d.%m.%Y"))
````

**Крок 0. Підготовка.** Запустіть клітинку нижче **один раз**. Після неї Python сам помічатиме, що Ви змінили свій модуль, і використовуватиме нову версію. Без неї зміни у файлі «не підхоплюються», доки не перезапустити блокнот.

````python
# ⚙️ Запустіть один раз перед роботою з модулями
%load_ext autoreload
%autoreload 2
````

**Крок 1. Створіть файл.** У Colab файл створює команда `%%writefile назва.py` у **першому рядку** клітинки. Ця клітинка код не виконує, а лише **зберігає його у файл**. Під клітинкою з'явиться `Writing money.py`, а сам файл — на панелі **Файли** 📁 ліворуч.

````python
%%writefile money.py
"""Функції для роботи з грошима."""


def fmt(amount):
    """Гарний вигляд суми: 1250.5 → '1250.50 грн'."""
    return f"{amount:.2f} грн"
````

## 5.3 Оглядово: що ще можна побачити в модулях

Детально ці речі розберемо пізніше. Зараз достатньо знати, що вони існують.

**1. Блок `if __name__ == "__main__":`**

Часто в кінці модуля трапляється такий запис:

```python
if __name__ == "__main__":
    print(fmt(1250.5))   # приклад роботи
```

Він означає: «виконати цей код **лише тоді**, коли файл запускають як окрему програму». Під час `import` цей код пропускається. Так у модуль додають приклади, які не заважають під час підключення.

**2. Структура проєкту.** Коли файлів стає багато, їх розкладають по папках:

```
sales_project/
├── main.py              ← головний файл з аналізом
├── analytics_utils.py   ← Ваші функції метрик
├── data/
│   └── orders.csv       ← дані
└── requirements.txt     ← список потрібних пакетів (тема 4)
```

**3. Як зберегти модуль.** Файли в Colab **зникають**, коли сесія закривається. Щоб не втратити модуль, завантажте його на комп'ютер (панель 📁 → ⋮ → *Завантажити*) або зберігайте на Google Диску. На власному комп'ютері достатньо покласти файл `.py` у ту саму папку, де лежить блокнот.

## ⚠️ Типові помилки

| Що бачите | Причина | Як виправити |
|---|---|---|
| `ModuleNotFoundError: No module named 'money'` | не запустили клітинку з `%%writefile` або помилилися в назві | запустіть клітинку з `%%writefile`, перевірте назву |
| код виконався замість запису у файл, або `UsageError` | `%%writefile` стоїть не в першому рядку | перенесіть `%%writefile` на самий перший рядок |
| `AttributeError: module 'money' has no attribute 'fmtt'` | помилка в назві функції | перевірте назву функції у файлі й у виклику |
| змінили файл, а результат старий | не запустили клітинку з `%%writefile` після змін | запустіть її ще раз (і перевірте, що виконано **Крок 0**) |

```python
def mean(values):
    return sum(values) / len(values)


def median(values):
    ordered = sorted(values)
    n = len(ordered)
    middle = n // 2
    if n % 2 == 1:
        return ordered[middle]
    return (ordered[middle - 1] + ordered[middle]) / 2


def value_range(values):
    return max(values) - min(values)


def describe(values):
    return {
        "count": len(values),
        "mean": round(mean(values), 2),
        "median": median(values),
        "range": value_range(values),
    }
```
