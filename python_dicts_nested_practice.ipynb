## 1. Dictionaries (Словники) — 10 завдань

### 📘 Теорія: пари ключ–значення

**Словник (`dict`)** зберігає дані парами **ключ → значення**. До значення звертаються не за номером, а за ключем: за назвою товару отримуємо ціну, за кодом клієнта — його дані.

```python
catalog = {"laptop": 32000, "mouse": 650}
#           ключ:   значення
```

| | Ключ | Значення |
|---|---|---|
| Унікальність | унікальний: повторений ключ перезапише попереднє значення | може повторюватися |
| Тип | тільки **незмінюваний**: рядок, число, кортеж | будь-який: число, рядок, список, інший словник |

**Способи створення:**
- фігурні дужки: `{"tea": 55, "coffee": 120}`;
- функція `dict()`: `dict(tea=55, coffee=120)`;
- зі списку пар: `dict([("tea", 55), ("coffee", 120)])`;
- порожній словник: `{}` або `dict()`.

````python
catalog = {"laptop": 32000, "mouse": 650, "monitor": 7800}
pairs = dict([("tea", 55), ("coffee", 120)])   # зі списку пар
empty = {}

print(catalog)        # {'laptop': 32000, 'mouse': 650, 'monitor': 7800}
print(pairs)          # {'tea': 55, 'coffee': 120}
print(len(catalog))   # 3 — кількість пар

dup = {"a": 1, "a": 2}
print(dup)            # {'a': 2} — повторений ключ перезаписав значення
````

### 📘 Доступ до значень: `[]` і `get()`

| | `d[key]` | `d.get(key)` / `d.get(key, default)` |
|---|---|---|
| Ключ є | повертає значення | повертає значення |
| Ключа немає | ❌ помилка `KeyError` | повертає `None` або `default` |
| Коли використовувати | ключ **обовʼязково має бути**; якщо його немає — це помилка в даних, і про неї краще дізнатися одразу | ключа **може не бути**, і це нормальна ситуація (наприклад, у товару немає знижки) |

````python
catalog = {"laptop": 32000, "mouse": 650, "monitor": 7800}

print(catalog["mouse"])              # 650
# catalog["keyboard"]  →  KeyError: 'keyboard'

print(catalog.get("mouse"))          # 650
print(catalog.get("keyboard"))       # None
print(catalog.get("keyboard", 0))    # 0
````

### 📘 `keys()`, `values()`, `items()`

| Метод | Повертає | Як побачити списком |
|---|---|---|
| `d.keys()` | усі ключі | `list(d.keys())` |
| `d.values()` | усі значення | `list(d.values())` |
| `d.items()` | пари `(ключ, значення)` — кортежі | `list(d.items())` |

- `key in d` перевіряє **ключі**; щоб шукати серед значень: `x in d.values()`.
- `sum()`, `min()`, `max()` застосовуйте до `d.values()`.

> ⚠️ `max(d)` поверне **найбільший ключ** (для рядків — останній за алфавітом), а не найбільше значення!

````python
catalog = {"laptop": 32000, "mouse": 650, "monitor": 7800}

print(catalog.keys())          # dict_keys(['laptop', 'mouse', 'monitor'])
print(list(catalog.values()))  # [32000, 650, 7800]
print(list(catalog.items()))   # [('laptop', 32000), ('mouse', 650), ('monitor', 7800)]

print("mouse" in catalog)          # True  — перевірка ключа
print(650 in catalog)              # False — 650 не є ключем
print(650 in catalog.values())     # True  — перевірка значення

print(max(catalog.values()))       # 32000 — найвища ціна
print(max(catalog))                # mouse — ⚠️ найбільший КЛЮЧ за алфавітом, а не ціна
print(sum(catalog.values()) / len(catalog))   # 13483.333333333334 — середня ціна
````

### 📘 Додавання, зміна, видалення

| Дія | Синтаксис | Примітка |
|---|---|---|
| Додати пару | `d[key] = value` | якщо ключа немає — додається |
| Змінити значення | `d[key] = value` | якщо ключ є — значення перезаписується |
| Змінити відносно поточного | `d[key] += 5`, `d[key] = d[key] * 1.1` | ключ має існувати, інакше `KeyError` |
| Оновити кілька пар | `d.update(other)` | змінює `d` на місці |
| Обʼєднати в новий словник | <code>d &#124; other</code> | `d` не змінюється |
| Видалити й отримати значення | `d.pop(key)` | `KeyError`, якщо ключа немає |
| Видалити «безпечно» | `d.pop(key, default)` | без помилки, поверне `default` |
| Видалити | `del d[key]` | |
| Очистити | `d.clear()` | |

> 💡 Для грошей зручно округлювати результат: `round(x, 2)` — два знаки після коми.

````python
catalog = {"laptop": 32000, "mouse": 650, "monitor": 7800}

catalog["keyboard"] = 1200          # нового ключа немає → додавання
catalog["mouse"] = 590              # ключ існує → заміна
catalog["laptop"] += 1000           # те саме, що catalog["laptop"] = catalog["laptop"] + 1000
print(catalog)   # {'laptop': 33000, 'mouse': 590, 'monitor': 7800, 'keyboard': 1200}

catalog["monitor"] = round(catalog["monitor"] * 0.85, 2)   # знижка 15%
print(catalog["monitor"])   # 6630.0

removed = catalog.pop("keyboard")
print(removed)                                        # 1200
print(catalog.pop("webcam", "немає такого товару"))   # немає такого товару

del catalog["mouse"]
print(catalog)   # {'laptop': 33000, 'monitor': 6630.0}

catalog.update({"monitor": 7000, "headphones": 2100})
print(catalog)   # {'laptop': 33000, 'monitor': 7000, 'headphones': 2100}

# catalog["webcam"] += 100  →  KeyError: 'webcam' (не можна збільшити те, чого немає)
````

### 📘 Практичний приклад: каталог товарів

Типовий сценарій: є каталог «товар → ціна», і його треба підтримувати в актуальному стані: додавати нові товари, змінювати ціни, прибирати те, що знято з продажу.

````python
# Каталог: назва товару → ціна, грн
catalog = {"notebook": 85, "pen": 25, "backpack": 1450}

# 1. Надійшов новий товар
catalog["marker"] = 60

# 2. Постачальник підняв ціну рюкзака на 8%
catalog["backpack"] = round(catalog["backpack"] * 1.08, 2)

# 3. Оновлюємо кілька цін одразу
catalog.update({"pen": 28, "notebook": 90})

# 4. Ручки знято з продажу
catalog.pop("pen")

print(catalog)                      # {'notebook': 90, 'backpack': 1566.0, 'marker': 60}
print("Товарів у каталозі:", len(catalog))   # 3
````

## 2. Nested structures & mutability (Вкладені структури і змінюваність) — 10 завдань

### 📘 Теорія: вкладені структури

Колекції можна вкладати одна в одну: список у словник, словник у список, словник у словник. Саме так виглядають реальні дані — відповіді API, JSON-файли, вивантаження з CRM.

| Структура | Приклад | Де трапляється |
|---|---|---|
| Список списків | `[[120, 95], [80, 110]]` | матриця, таблиця без назв колонок |
| Словник зі списками | `{"Mon": ["SQL", "Python"]}` | розклад, теги, товари в замовленні |
| Словник словників | `{"C001": {"name": "Anna"}}` | довідник клієнтів за ID |
| **Список словників** | `[{"name": "Anna"}, {"name": "Taras"}]` | **таблиця**: кожен словник — рядок |

**Доступ до вкладених елементів** — ланцюжок квадратних дужок, який читається зліва направо. Кожен крок повертає обʼєкт, до якого застосовується наступний:

```
orders[0]["items"][1]
  │      │        └─ другий елемент списку товарів
  │      └────────── значення ключа "items" (список)
  └───────────────── перше замовлення (словник)
```

На кожному кроці памʼятайте, з чим ви працюєте: до **списку** звертаються за індексом-числом, до **словника** — за ключем.

````python
# Список списків: продажі 3 магазинів за 2 дні
sales = [[120, 95], [80, 110], [150, 60]]
print(sales[2][0])               # 150 — третій магазин, перший день

# Словник зі списками у значеннях
schedule = {"Mon": ["SQL", "Python"], "Tue": ["Excel"]}
print(schedule["Mon"][1])        # Python

# Словник словників
clients = {
    "C001": {"name": "Anna", "city": "Kyiv"},
    "C002": {"name": "Taras", "city": "Lviv"},
}
print(clients["C002"]["city"])   # Lviv
````

### 📘 Список словників як «таблиця»

Кожен словник — це **рядок**, ключі — **назви колонок**:

| індекс | name | dept | salary |
|---|---|---|---|
| 0 | Anna | Sales | 32000 |
| 1 | Bohdan | Marketing | 28000 |
| 2 | Olena | Sales | 35000 |

- `table[i]` — весь рядок (словник);
- `table[i]["col"]` — одна «клітинка»;
- `table.append({...})` — додати рядок;
- `table[i]["col"] = value` — змінити клітинку.

> 💡 Отримати всю колонку одразу (наприклад, усі зарплати) простими засобами поки не вийде — для цього знадобляться цикли, а згодом бібліотека pandas, де така «таблиця» перетворюється на DataFrame одним рядком коду.

````python
employees = [
    {"name": "Anna",   "dept": "Sales",     "salary": 32000},
    {"name": "Bohdan", "dept": "Marketing", "salary": 28000},
    {"name": "Olena",  "dept": "Sales",     "salary": 35000},
]

print(len(employees))              # 3 — кількість рядків
print(employees[1])                # {'name': 'Bohdan', 'dept': 'Marketing', 'salary': 28000}
print(employees[2]["salary"])      # 35000 — «клітинка»: рядок 2, колонка salary
print(list(employees[0].keys()))   # ['name', 'dept', 'salary'] — «назви колонок»

employees[1]["salary"] = 30000                                      # змінити клітинку
employees.append({"name": "Iryna", "dept": "IT", "salary": 41000})  # додати рядок
print(len(employees))              # 4
print(employees[-1]["name"])       # Iryna
````

### 📘 Змінювані й незмінювані типи

| Незмінювані (immutable) | Змінювані (mutable) |
|---|---|
| `int`, `float`, `bool`, `str`, `tuple` | `list`, `dict`, `set` |
| будь-яка «зміна» створює **новий** обʼєкт | змінюються **на місці** |
| можуть бути ключами словника й елементами множини | не можуть бути ключами словника й елементами множини |

Наслідки, які варто памʼятати:
- методи рядків (`upper()`, `replace()`, `strip()`) **повертають новий рядок**, а не змінюють старий — результат треба записати в змінну;
- методи списків (`append()`, `sort()`) **змінюють сам список** і повертають `None`;
- кортеж незмінюваний, але якщо в ньому лежить список, **сам список** змінити можна.

````python
city = "kyiv"
city.upper()
print(city)            # kyiv — рядок не змінився
city = city.upper()    # щоб «змінити», записуємо новий рядок у змінну
print(city)            # KYIV

nums = [3, 1, 2]
nums.sort()
print(nums)            # [1, 2, 3] — список змінився на місці

record = ("Anna", [90, 85])
record[1].append(70)   # кортеж не змінюємо — змінюємо список усередині нього
print(record)          # ('Anna', [90, 85, 70])
# record[0] = "Olena"  →  TypeError: 'tuple' object does not support item assignment
````

### 📘 Посилання: `=` не створює копію

Змінна в Python — це **ярлик**, приклеєний до обʼєкта. Запис `b = a` не копіює дані, а приклеює другий ярлик до того самого обʼєкта.

- `a == b` — чи однаковий **вміст**;
- `a is b` — чи це **той самий обʼєкт**.

Для незмінюваних типів це непомітно: «змінити» число чи рядок неможливо, тому завжди створюється новий обʼєкт. А от для списків, словників і множин зміна через один ярлик видна через усі інші.

````python
a = [1, 2, 3]
b = a                  # b — ще один ярлик на ТОЙ САМИЙ список
b.append(4)
print(a)               # [1, 2, 3, 4] — змінився і a
print(a is b)          # True — це один обʼєкт

c = [1, 2, 3, 4]
print(a == c)          # True  — вміст однаковий
print(a is c)          # False — але це різні обʼєкти

x = 10
y = x
y = y + 5              # для чисел нове значення — це новий обʼєкт
print(x, y)            # 10 15
````

### 📘 `copy` і `deepcopy`

| Спосіб | Що копіює |
|---|---|
| `b = a` | нічого — це той самий обʼєкт |
| **Поверхнева копія:** `a.copy()`, `list(a)`, `a[:]`, `dict(a)`, `copy.copy(a)` | створює новий **зовнішній** контейнер, але вкладені обʼєкти залишаються **спільними** |
| **Глибока копія:** `copy.deepcopy(a)` | копіює **все**, на всіх рівнях вкладеності |

`deepcopy` знаходиться в модулі `copy`, тому спочатку його треба підключити: `import copy`.

**Правило:**
- колекція з незмінюваних елементів (числа, рядки) → вистачить `.copy()`;
- колекція, всередині якої є списки, словники, множини → потрібен `copy.deepcopy()`.

````python
import copy

order = {"id": 1, "items": ["milk", "bread"]}

shallow = order.copy()           # поверхнева копія
deep = copy.deepcopy(order)      # глибока копія

shallow["id"] = 99               # верхній рівень у копії незалежний
shallow["items"].append("eggs")  # а вкладений список — СПІЛЬНИЙ з оригіналом!

print(order)     # {'id': 1, 'items': ['milk', 'bread', 'eggs']}  ← оригінал зачепило
print(shallow)   # {'id': 99, 'items': ['milk', 'bread', 'eggs']}
print(deep)      # {'id': 1, 'items': ['milk', 'bread']}          ← повністю незалежна
````

### 📘 Практичний приклад: замовлення як список словників

Кожне замовлення — словник, а товари в ньому — вкладений список. Вибираємо потрібні значення ланцюжками індексів і ключів.

````python
orders = [
    {"id": 101, "client": "Anna",  "items": ["laptop", "mouse"],             "total": 32650, "status": "paid"},
    {"id": 102, "client": "Taras", "items": ["monitor"],                     "total": 7800,  "status": "new"},
    {"id": 103, "client": "Anna",  "items": ["keyboard", "mouse", "pad"],    "total": 2400,  "status": "shipped"},
]

print(orders[1]["client"])        # Taras    — клієнт другого замовлення
print(orders[2]["items"][0])      # keyboard — перший товар третього замовлення
print(len(orders[2]["items"]))    # 3        — кількість товарів у третьому замовленні
print(orders[0]["items"][-1])     # mouse    — останній товар першого замовлення

orders[1]["status"] = "paid"               # змінюємо статус
orders[1]["items"].append("hdmi cable")    # додаємо товар до замовлення
print(orders[1])

clients = {orders[0]["client"], orders[1]["client"], orders[2]["client"]}
print(len(clients))                                                    # 2 — унікальних клієнтів
print(orders[0]["total"] + orders[1]["total"] + orders[2]["total"])   # 42850 — загальна сума
````

### 📘 Повторення: шпаргалка по чотирьох структурах

| Дія | `list` | `tuple` | `set` | `dict` |
|---|---|---|---|---|
| Створити | `[1, 2]` | `(1, 2)` | `{1, 2}` | `{"a": 1}` |

| Змінюваний | ✅ | ❌ | ✅ | ✅ |

### 📘 Як обрати структуру даних під задачу

Поставте собі кілька запитань:

1. **Чи важлива унікальність** або треба порівнювати набори (спільне, відмінне)? → `set`
2. **Чи треба знаходити значення за назвою, кодом, ідентифікатором?** → `dict`
3. **Чи це фіксований запис, який не має змінюватися** (або має бути ключем словника)? → `tuple`
4. **Чи це впорядкована послідовність, яку будемо доповнювати й змінювати?** → `list`
5. **Чи це багато записів з однаковими полями?** → список словників

| Задача | Структура | Чому |
|---|---|---|
| Щоденна виручка за місяць | `list` | важливий порядок днів, значення можуть повторюватися |
| Координати складу | `tuple` | фіксована пара значень |
| Email-адреси підписників | `set` | без повторів, швидка перевірка «чи вже підписаний» |
| Ціна за артикулом | `dict` | шукаємо ціну за ключем |
| Залишки по товарах | `dict` | товар → кількість, кількість часто змінюється |
| Журнал продажів (дата, товар, сума) | список словників | багато записів з однаковою структурою |

> 💡 Часто одна задача потребує **кількох** структур одночасно — як у міні-задачі в кінці цього розділу.

### 📘 Типові помилки

| Помилка | Що побачите | Як правильно |
|---|---|---|
| Неіснуючий ключ: `d["x"]` | `KeyError: 'x'` | `d.get("x", default)` або спочатку `"x" in d` |
| Індекс за межами списку | `IndexError: list index out of range` | перевірити `len()`; для останнього елемента — `[-1]` |
| `{}` замість порожньої множини | `AttributeError: 'dict' object has no attribute 'add'` | `set()` |
| `lst = lst.append(x)` | у змінній опиняється `None` | `lst.append(x)` без присвоєння |
| `("a")` замість кортежу | отримуєте рядок, а не кортеж | `("a",)` — з комою |
| Індекс у множині: `s[0]` | `TypeError: 'set' object is not subscriptable` | перевірка `x in s` або `list(s)` |
| Список як ключ словника / елемент множини | `TypeError: unhashable type: 'list'` | використати кортеж |
| `x in d`, щоб знайти значення | `False`, хоча значення є | `x in d.values()` |
| `max(d)` для найбільшого значення | найбільший **ключ** | `max(d.values())` |
| `b = a` як копія | зміни в `b` змінюють `a` | `a.copy()` |
| `.copy()` для вкладеної структури | зміни вкладених обʼєктів видно в оригіналі | `copy.deepcopy(a)` |
| Зміна елемента кортежу | `TypeError: 'tuple' object does not support item assignment` | створити новий кортеж або працювати зі списком |

> 💡 **Як читати повідомлення про помилку:** дивіться на **останній рядок**. Спочатку там тип помилки (`KeyError`, `TypeError`…), потім пояснення. Вище стрілкою `---->` позначено рядок коду, де вона сталася.
