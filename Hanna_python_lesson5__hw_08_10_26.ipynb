10. Створи три змінні на основі `students`:
- `ref` — просте присвоєння  (b = a);
- `shallow` — поверхнева копія (b = copy(a));
- `deep` — глибока копія (b = copy.deepcopy(a)).

Потім у `shallow` додай студентці `"Sofia"` оцінку 100, а з `ref` видали останнього студента.

Виведи кількість студентів та оцінки студентки `"Sofia"` в `students`, `ref`, `shallow` і `deep`. Поясни в коментарі отримані результати

````python
import copy

ref = students          #ref це не копія, а друга назва того самого списку, тому len(students) == len(ref) і видалення з ref зменшує і students.
shallow = copy.copy(students)     # shallow — новий список, але всередині лежать ті самі словники, тому append до scores відображається і в students, і в ref; але видалення елемента зі students на довжину shallow не впливає.
deep = copy.deepcopy(students)    # deep — глибока копія і списка, і словників, і вкладеного списку оцінок, тому його не зачепила ні зміна оцінок, ні видалення студента.
shallow[2]["scores"].append(100)
del ref[-1]
print(f'Students: {len(students)}, {students[2].get("scores")}')
print(f'Ref: {len(ref)}, {ref[2].get("scores")}')
print(f'Shallow: {len(shallow)}, {shallow[2].get("scores")}')
print(f'Deep: {len(deep)}, {deep[2].get("scores")}')
````

### 📚 Міні-задача: облік бібліотеки

Дані:
- `books` — книги в наявності: назва → кількість примірників;
- `borrowed_today` — книги, видані сьогодні: кожен елемент — один виданий примірник;
- `readers_mon`, `readers_tue` — читачі, які приходили в понеділок і у вівторок.

Запусти клітинку з даними, а потім виконуй завдання 14 і 15

````python
books = {"1984": 3, "Dune": 2, "Atomic Habits": 4, "Sapiens": 1}
borrowed_today = ["Dune", "1984", "Sapiens", "1984"]
readers_mon = {"Iryna", "Maksym", "Oleh"}
readers_tue = {"Maksym", "Sofia"}
````

14. Збережи стан `books` до видачі у змінну `books_before` так, щоб подальші зміни `books` на неї не вплинули.  
Зменш кількість примірників кожної книги на кількість виданих сьогодні.  
Виведи `books_before` і `books`, а також множину книг, які сьогодні не видавалися

````python
books_before = books.copy()

for book in borrowed_today:
  books[book] -= 1

print(f'До видачі: {books_before}')
print(f'Після видачі: {books}')
print(f'Не видавалися сьогодні: {set(books_before) - set(borrowed_today)}')
````

15. Збери звіт у словник `report` з ключами:
- `"titles"` — скільки різних назв книг є в бібліотеці;
- `"copies_left"` — скільки примірників залишилося загалом;
- `"out_of_stock"` — чи є книга, якої не залишилося в наявності (`True` / `False`);
- `"unique_readers"` — скільки різних читачів приходило за два дні;
- `"loyal_readers"` — множина читачів, які приходили обидва дні.

Виведи `report`

````python
report = {
  "titles": len(books),
  "copies_left": sum(books.values()),
  "out_of_stock": 0 in books.values(),
  "unique_readers": len(readers_mon | readers_tue),
  "loyal_readers": readers_mon & readers_tue
}
print(report)
````
