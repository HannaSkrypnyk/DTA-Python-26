2. Відбери курси з категорії `"data"` з рейтингом 4.5 і вище. Для відібраних курсів отримай:
- кількість таких курсів;
- список їхніх назв великими літерами;
- список цін зі знижкою 15%, округлених до цілих.

Розвʼяжи завдання за допомогою `filter` і `map`

```python
# filter() лишає елементи, для яких lambda повертає True"
count = 0
list_name = []
list_price = []
for course in filter(lambda x: x['category'] == 'data', COURSES):
  if course['rating'] >= 4.5:
    count += 1
    list_name.append(course['title'])
    list_price.append(round(course['price'] * 0.85))
  print(f"{course['title'].upper()} - {course['price']}, count - {count}")

```

```python
data_top = list(filter(lambda course: course["category"] == "data" and course["rating"] >= 4.5, COURSES))

# map() застосовує lambda до кожного відібраного курсу"
titles_upper = list(map(lambda course: course["title"].upper(), data_top))
prices_discounted = list(map(lambda course: round(course["price"] * 0.85), data_top))

print(f"Кількість курсів: {len(data_top)}")
print(f"Назви: {titles_upper}")
print(f"Ціни зі знижкою 15%: {prices_discounted}")
```

3. Напиши функцію `top_by(items, field, n=3, condition=None)` з docstring, яка повертає `n` елементів з найбільшим значенням поля `field`. Якщо передано `condition` — функцію-умову, — спершу залиш лише ті елементи, для яких вона повертає `True`.

Виклич функцію й виведи назви курсів:
- 3 курси з найвищим рейтингом;
- 2 найпопулярніші курси (за кількістю студентів) серед курсів категорії `"bi"`;
- 1 найдовший курс (за кількістю годин)

```python
def top_by(items, field, n=3, condition=None):
  """Повертає n елементів списку з найбільшим значенням поля field.
  Аргументи:  items (list[dict]): список словників, серед яких шукаємо.
              field (str): назва поля, за яким порівнюємо елементи.
              n (int): скільки елементів повернути (за замовчуванням 3).
              condition (callable | None): функція-умова; якщо задана, спершу лишаються лише ті елементи, для яких вона повертає True.
  Повертає:  list[dict]: до n елементів, відсортованих за спаданням поля field.    """

  if condition is None:
    selected = items
  else:
    selected = list(filter(condition, items))
  return sorted(selected, key=lambda item: item[field], reverse=True)[:n]

print("Топ-3 за рейтингом:")
for course in top_by(COURSES, "rating"):
  print(" ", course["title"], course["rating"])
print()

print("Топ-2 BI за кількістю студентів:")
for course in top_by(COURSES, "students", n=2, condition=lambda c: c["category"] == "bi"):
  print(" ", course["title"], course["students"])

print("\nНайдовший курс:")
for course in top_by(COURSES, "hours", n=1):
  print(" ", course["title"], course["hours"], "год")
```

4. Служба доставки бере 35 грн за кожен **розпочатий** кілограм ваги посилки: посилка вагою 1.2 кг оплачується як 2 кг.  
Дано ваги посилок у кілограмах: `weights = [0.4, 2.0, 3.7, 1.2, 12.5, 0.9]`.

Виведи вартість доставки кожної посилки та загальну вартість. Окремо виведи загальну вагу всіх посилок, округлену **вниз** до цілих кілограмів

```python
import math

weights = [0.4, 2.0, 3.7, 1.2, 12.5, 0.9]

price_per_kg = 35
total_cost = 2

for i in weights:
  cost = math.ceil(i) * price_per_kg
  total_cost += cost
  print(f'{i} кг, вартість доставки - {cost} грн.')

print(f'\nЗагальна вартість доставки: {total_cost} грн.')
print(f'Загальна вага: {math.floor(sum(weights))} кг')
```

6. Згенеруй тестові дані для 10 замовлень. Встанови `seed` рівним 2026. Для кожного замовлення спочатку згенеруй суму — ціле число від 100 до 5000 включно, — а потім випадково обери спосіб оплати зі списку `["card", "cash", "apple pay"]`. Збережи замовлення як список кортежів `(сума, спосіб оплати)`.

Виведи:
- усі згенеровані замовлення;
- кількість замовлень кожним способом оплати;
- частку замовлень на суму понад 2000 грн у відсотках;
- 2 різних випадкових замовлення для перевірки якості, вибрані з уже згенерованого списку

```python
import random

random.seed(2026)
payment_method = ["card", "cash", "apple pay"]
orders = []

for i in range(10):
  amount = random.randint(100, 5000)
  method = random.choice(payment_method)
  orders.append((amount, method))

print("Згенеровані замовлення:")
for amount, method in orders:
  print(f" {amount:>5} грн — {method}")

print("\nКількість замовлень за способом оплати:")
for method in payment_method:
  count = sum(1 for amount, pay in orders if pay == method)
  print(f" {method:<10} {count}")

big_orders = [order for order in orders if order[0] > 2000]
print(f"\nЗамовлень понад 2000 грн: {len(big_orders)} з {len(orders)} "
      f"= {len(big_orders) / len(orders) * 100:.1f}%")

print("\n2 випадкові замовлення для перевірки якості:", random.sample(orders, 2))
```

7. **Аналіз підписок.** Дано підписки клієнтів у форматі `(клієнт, дата початку, тривалість у днях)` і дату, на яку робиться звіт:

`TODAY = "2026-10-21"`  
`subscriptions = [("anna", "2026-09-01", 30), ("bohdan", "2026-10-01", 30), ("iryna", "2026-08-15", 90), ("oleh", "2026-10-10", 14), ("sofia", "2026-07-01", 365), ("taras", "2026-10-15", 7)]`

Для кожної підписки визнач:
- дату закінчення у форматі `ДД.ММ.РРРР`;
- день тижня, на який припадає дата закінчення (англійською, наприклад `Saturday`);
- скільки днів залишилося до закінчення від `TODAY` (для завершених підписок — відʼємне число);
- статус: `"expired"` — підписка вже закінчилася, `"expires soon"` — залишилося 7 днів або менше, `"active"` — в інших випадках.

Виведи результат таблицею за допомогою пакета `tabulate`. Наприкінці виведи кількість підписок кожного статусу та медіану кількості днів, що залишилися, для активних підписок

```python
!pip install tabulate -q
import random
from datetime import date, timedelta
import statistics
from tabulate import tabulate
from collections import Counter

TODAY = "2026-10-21"
subscriptions = [
    ("anna", "2026-09-01", 30),
    ("bohdan", "2026-10-01", 30),
    ("iryna", "2026-08-15", 90),
    ("oleh", "2026-10-10", 14),
    ("sofia", "2026-07-01", 365),
    ("taras", "2026-10-15", 7)
  ]

today = date.fromisoformat(TODAY)        # рядок "2026-10-30"

rows = []
for client, start_str, duration in subscriptions:
  end = date.fromisoformat(start_str) + timedelta(days=duration)   # дата закінчення
  days_left = (end - today).days

  if days_left < 0:
    status = "expired"
  elif days_left <= 7:
    status = "expires soon"
  else:
    status = "active"

  rows.append([client, end.strftime("%d.%m.%Y"), end.strftime("%A"), days_left, status])

headers = ["клієнт", "дата закінчення", "день тижня", "днів залишилось", "статус"]
print(tabulate(rows, headers=headers, tablefmt="grid"))

print("\nКількість підписок за статусом:")
for status in ["expired", "expires soon", "active"]:
  print(status, sum(1 for row in rows if row[4] == status))

active_days = [row[3] for row in rows if row[4] == "active"]
print(f"\nМедіана днів для активних підписок: {statistics.median(active_days)}")
```
